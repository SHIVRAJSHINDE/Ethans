<p style="color: #2a9d8f; font-family: 'Candara', sans-serif; font-size: 16px; margin: 0; padding: 0;">

DataFrames are incredibly versatile and are widely used for data manipulation, analysis, and exploration in Python.

<p style="color: #264653; font-family: 'Candara', sans-serif; font-size: 18px; line-height: 1; margin-top: 0;">
Handling missing values is essential in data cleaning. Pandas provides two primary methods: <code>fillna</code> and <code>dropna</code>.
</p>

<ul style="font-family: 'Candara', sans-serif; font-size: 18px; line-height: 1; padding-left: 20px; margin-top: 0; list-style-type: none;">
    <li style="color: #2a9d8f; margin-bottom: 0;"><b>.fillna</b>
        <ul style="margin: 0; padding-left: 15px; list-style-type: none;">
            <li style="margin-bottom: 0;"><code>df.fillna(0)</code> — Replace all NaNs with 0.</li>
            <li style="margin-bottom: 0;"><code>df.fillna(method='ffill')</code> — Forward fill using last valid value.</li>
            <li style="margin-bottom: 0;"><code>df.fillna(method='bfill')</code> — Backward fill using next valid value.</li>
            <li style="margin-bottom: 0;"><code>df.fillna({'col1':0,'col2':1})</code> — Fill NaNs in 'col1' and 'col2' with 0 and 1.</li>
        </ul>
    </li>
    <li style="color: #2a9d8f; margin-bottom: 0;"><b>.dropna</b>
        <ul style="margin: 0; padding-left: 15px; list-style-type: none;">
            <li style="margin-bottom: 0;"><code>df.dropna()</code> — Drop rows where any value is NaN.</li>
            <li style="margin-bottom: 0;"><code>df.dropna(axis=1)</code> — Drop columns where any value is NaN.</li>
            <li style="margin-bottom: 0;"><code>df.dropna(how='all')</code> — Drop rows where all values are NaN.</li>
            <li style="margin-bottom: 0;"><code>df.dropna(subset=['col1','col2'])</code> — Drop rows where NaNs appear in 'col1' or 'col2'.</li>
        </ul>
    </li>
</ul>

<p style="color: #264653; font-family: 'Candara', sans-serif; font-size: 18px; line-height: 1; margin-top: 0;">
The <code>dropna</code> method removes missing values, while <code>fillna</code> fills missing values in a DataFrame or Series with a specified value or method.

<h2 style="color: #2a9d8f; font-family: 'Candara', sans-serif; font-weight: bold; text-align: left;margin:-60px 0 0 0; padding-bottom: 10px; border-bottom: 2px solid #264653;">  Combining DataFrames – Merging Operatio 
</h2>ft.</li>
</ul>


<p style="color: #264653; font-family: 'Candara', sans-serif; font-size: 18px; line-height: 1; margin-top: 0;">
Merge two DataFrames based on a key column using the <code>pd.merge()</code> function.
</p>

<h3 style="color: #2a9d8f; font-family: 'Candara', sans-serif; margin-top: 0;">Syntax</h3>
<pre style="line-height: 1;">pd.merge(df1, df2, left_on='name1', right_on='name2', how='outer')</pre>

<h3 style="color: #2a9d8f; font-family: 'Candara', sans-serif; margin-top: 0;">Join Types</h3>
<ul style="font-family: 'Candara', sans-serif; font-size: 18px; line-height: 1; padding-left: 20px; margin-top: 0; list-style-type: none;">
    <li style="color: #2a9d8f; margin-bottom: 0;"><code>how='outer'</code> — Returns all rows from both DataFrames, filling missing values with NaN.</li>
    <li style="color: #2a9d8f; margin-bottom: 0;"><code>how='inner'</code> — Returns only rows with matching keys in both DataFrames.</li>
    <li style="color: #2a9d8f; margin-bottom: 0;"><code>how='left'</code> — Returns all rows from left DataFrame, with matching rows from right.</li>
    <li style="color: #2a9d8f; margin-bottom: 0;"><code>how='right'</code> — Returns all rows from right DataFrame, with matching rows from left.</li>
