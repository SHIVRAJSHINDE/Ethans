<h2 style="color: #2a9d8f; font-family: 'Candara', sans-serif; font-weight: bold; text-align: left; padding-bottom: 10px; border-bottom: 2px solid #264653;">Iteration over DataFrame</h2>

<p style="color: #264653; font-family: 'Candara', sans-serif; font-size: 12px; line-height: 1.5; margin-top: 20px;">
    Iterating over a DataFrame can be done in several ways depending on your needs. Here are some common methods:
</p>


<ol style="font-family: 'Candara', sans-serif; font-size: 12px; line-height: 1.5; padding-left: 20px; margin-top: 10px; list-style-type: none;">
    <li style="color: #2a9d8f; margin-bottom: 5px;">
        <b>Using <code>iterrows()</code></b>
        <p style="color: #2a9d8f; font-family: 'Candara', sans-serif; font-size: 15px; margin: 0; line-height: 1.2;">
            Iterate over DataFrame rows as (index, Series) pairs.<br>
            <i>Syntax:</i> <code>for index, row in df.iterrows():</code>
        </p>
    </li>
    <li style="color: #2a9d8f; margin-bottom: 5px;">
        <b>Using <code>itertuples()</code></b>
        <p style="color: #2a9d8f; font-family: 'Candara', sans-serif; font-size: 15px; margin: 0; line-height: 1.2;">
            Iterate over DataFrame rows as named tuples (Generally faster than <code>iterrows()</code>).<br>
            <i>Syntax:</i> <code>for row in df.itertuples():</code>
        </p>
    </li>
    <li style="color: #2a9d8f;">
        <b>Using <code>items()</code></b>
        <p style="color: #2a9d8f; font-family: 'Candara', sans-serif; font-size: 15px; margin: 0; line-height: 1.2;">
            Iterate over DataFrame columns as (column\_name, Series) pairs.<br>
            <i>Syntax:</i> <code>for column_name, series in df.items():</code>
        </p>

<h2 style="color: #2a9d8f; font-family: 'Candara', sans-serif; font-weight: bold; text-align: left; padding-bottom: 10px; border-bottom: 2px solid #264653;">Use of Map, Apply and ApplyMap</h2>

<p style="color: #264653; font-family: 'Candara', sans-serif; font-size: 15px; line-height: 1.5; margin-top: 15px;">
These functions allow for the application of logic (functions or mappings) to Series or DataFrames efficiently.
</p>

<ol style="font-family: 'Candara', sans-serif; font-size: 15px; line-height: 1.5; padding-left: 20px; margin-top: 10px; list-style-type: none;">

<li style="color: #2a9d8f; margin-bottom: 10px;">
<b>Using <code>map()</code></b>
<p style="color: #264653; font-family: 'Candara', sans-serif; font-size: 14px; margin: 0; line-height: 1.2;">
Map values of a Series using a function or dict. **(Only works on a Series/Column)**
</p>
<p style="color: #2a9d8f; font-family: 'Candara', sans-serif; font-size: 14px; margin: 0; line-height: 1.2;">
<i>Syntax (Function):</i> <code>df['column'].map(func)</code><br>
<i>Syntax (Dictionary):</i> <code>df['column'].map(dict)</code>
</p>

</li>

<li style="color: #2a9d8f; margin-bottom: 10px;">
<b>Using <code>apply()</code></b>
<p style="color: #264653; font-family: 'Candara', sans-serif; font-size: 14px; margin: 0; line-height: 1.2;">
Apply a function along an axis of the DataFrame (column or row).
</p>
<p style="color: #2a9d8f; font-family: 'Candara', sans-serif; font-size: 14px; margin: 0; line-height: 1.2;">
<i>Syntax (Single Column):</i> <code>df['ColName'].apply(func)</code><br>
<i>Syntax (On Columns/Axis 0):</i> <code>df.apply(func, axis=0)</code><br>
<i>Syntax (On Rows/Axis 1):</i> <code>df.apply(func, axis=1)</code>
</p>

</li>

<li style="color: #2a9d8f;">
<b>Using <code>applymap()</code></b>
<p style="color: #264653; font-family: 'Candara', sans-serif; font-size: 14px; margin: 0; line-height: 1.2;">
Apply a function element-wise across the entire DataFrame.
</p>
<p style="color: #2a9d8f; font-family: 'Candara', sans-serif; font-size: 14px; margin: 0; line-height: 1.2;">
<i>Syntax:</i> <code>df.applymap(func)</code>
</p>

</li>

<h2 style="color: #2a9d8f; font-family: 'Candara', sans-serif; font-weight: bold; text-align: left; padding-bottom: 10px; border-bottom: 2px solid #264653;">
Concatenation, Joining, and Merging Operations in Pandas
</h2>

<p style="color: #264653; font-family: 'Candara', sans-serif; font-size: 15px; line-height: 1.5; margin-top: 15px;">
Combining DataFrames can be achieved through concatenation, joining, and merging. Here are the common methods:
</p>

<ol style="font-family: 'Candara', sans-serif; font-size: 15px; line-height: 1.5; padding-left: 20px; margin-top: 10px; list-style-type: none;">

# <h2 style="color: #2a9d8f; font-family: 'Candara', sans-serif; font-weight: bold; text-align: center; padding-bottom: 10px; border-bottom: 2px solid #264653;">Concatenation in Pandas</h2>

<p style="color: #264653; font-family: 'Candara', sans-serif; font-size: 18px; line-height: 1.8; margin-top: 20px;">
    Concatenation is used to combine DataFrames along a specific axis. Here are the common methods for concatenating DataFrames:
</p>

<ol style="font-family: 'Candara', sans-serif; font-size: 18px; line-height: 1.8; padding-left: 20px; margin-top: 10px; list-style-type: noon</b>
        <p style="color: #2a9d8f; font-family: 'Candara', sans-serif; font-size: 16px; margin: 0;">
            To concatenate DataFrames along rows: <code>pd.concat([df1, df2, df3], axis=0)</code><br>
            To concatenate DataFrames along columns: <code>pd.concat([df1, df2, df3], axis=1)</code>
        </p>
    </li>
</ol>
    </li>
</ol>


```python
import pandas as pd

df1 = pd.DataFrame(
        {"A": ["A0", "A1", "A2", "A3"],
            "B": ["B0", "B1", "B2", "B3"],
        },
        index=[0, 1, 2, 3],
    )
    

df2 = pd.DataFrame(
        {   "A": ["A0", "A1", "A2", "A3"],
            "B": ["B4", "B5", "B6", "B7"],
                 },
        index=[4, 5, 6, 7],
    )
    

df3 = pd.DataFrame({"A": ["A0", "A1", "A2", "A3"],
                    "B": ["B8", "B9", "B10", "B11"],},
                    index=[8, 9, 10, 11],)

print(df1)
print()
print(df2)
print()
print(df3)
```

```python
# Concant dataframe can be joined vertically or horizontally but no outer, inner, left or right join are possible
```

```python
'''Join Dataframe on top of the above '''
allDf = [df1,df2,df3]
# row = 0
# column = 1
pd.concat(allDf,axis=0)
```

```python
'''Join Dataframe on top of the above '''
allDf = [df1,df2,df3]
# row = 0
# column = 1
pd.concat(allDf,axis=1)
```

```python
'''Join Dataframe Side by Side '''
# here common key is not considered
allDf = [df1, df2, df3]

result_df = pd.concat(allDf, axis=1)
result_df
```

# <h2 style="color: #2a9d8f; font-family: 'Candara', sans-serif; font-weight: bold; text-align: center; padding-bottom: 10px; border-bottom: 2px solid #264653;">Join Types in Pandas</h2>

<p style="color: #264653; font-family: 'Candara', sans-serif; font-size: 18px; line-height: 1.8; margin-top: 20px;">
    Different join types determine which rows are included in the result:
</p>

<ul style="font-family: 'Candara', sans-serif; font-size: 18px; line-height: 1.8; padding-left: 20px; margin-top: 10px; list-style-type: none;">
    <li style="color: #2a9d8f;"><b>Outer Join</b>: Includes all rows from both DataFrames.<br>
        <code>how='outer'</code>
    </li>
    <li style="color: #2a9d8f;"><b>Inner Join</b>: Includes only rows with matching keys in both DataFrames.<br>
        <code>how='inner'</code>
    </li>
    <li style="color: #2a9d8f;"><b>Left Join</b>: Includes all rows from the left DataFrame and matching rows from the right.<br>
        <code>how='left'</code>
    </li>
    <li style="color: #2a9d8f;"><b>Right Join</b>: Includes all rows from the right DataFrame and matching rows from the left.<br>
        <code>how='right'</code>
    </li>
</ul>
li>
</ul>


# <h2 style="color: #2a9d8f; font-family: 'Candara', sans-serif; font-weight: bold; text-align: center; padding-bottom: 10px; border-bottom: 2px solid #264653;">Merging DataFrames in Pandas</h2>

<p style="color: #264653; font-family: 'Candara', sans-serif; font-size: 18px; line-height: 1.8; margin-top: 20px;">
    Merging DataFrames combines them based on common columns or indices. Here’s how different join types work:
</p>

<ol style="font-family: 'Candara', sans-serif; font-size: 18px; line-height: 1.8; padding-left: 20px; margin-top: 10px; list-style-type: none;">
    <li style="color: #2a9d8f;"><b>Outer Join</b>
        <p style="color: #2a9d8f; font-family: 'Candara', sans-serif; font-size: 16px; margin: 0;">
            Includes all rows from both DataFrames, with `NaN` for non-matching rows.<br>
            <code>pd.merge(df1, df2, left_on='name1', right_on='name2', how='outer')</code>
        </p>
</ol>


```python
#  Create DataFrame

data1 = [("Alice", 1),("Charlie", 2),("Bob", 3)]
columns1 = ["name1", "value1"]

df1 = pd.DataFrame(data1, columns=columns1)
print(df1)

print()

data2 = [("Alice", 10), ("Bob", 20), ("David", 30)]
columns2 = ["name2", "value2"]
df2 = pd.DataFrame(data2, columns=columns2)
print(df2)

```

```python
# outer = 
# inner
# left
# right
```

```python
# merge()
# Outer join
# in merge method common_key is considered

# pd.merge(df1, df2, left_on='NameOfCol', right_on='NameOfCol', how='outer')

pd.merge(df1,df2,left_on='name1',right_on='name2',how='outer')

```

```python
# inner join
# in merge method common_key is considered

merged_df = pd.merge(df1, df2, left_on='name1', right_on='name2',how='inner')
merged_df
```

```python
# left join
# in merge method common_key is considered

merged_df = pd.merge(df1, df2, left_on='name1', right_on='name2',how='left')
merged_df
```

```python
# right join
# in merge method common_key is considered

merged_df = pd.merge(df1, df2, left_on='name1', right_on='name2',how='right')
merged_df
```

<h2 style="color: #2a9d8f; font-family: 'Candara', sans-serif; font-weight: bold; text-align: left; padding-bottom: 10px; border-bottom: 2px solid #264653;">
Pandas GroupBy
</h2>

<p style="color: #264653; font-family: 'Candara', sans-serif; font-size: 15px; line-height:1; margin-top: 0;">
The <code>groupby</code> method in Pandas is used to group data based on one or more columns and perform operations on these groups. It is useful for aggregating and summarizing data.
</p>

<ol style="font-family: 'Candara', sans-serif; font-size: 15px; line-height:1; padding-left: 20px; margin-top: 0; list-style-type: none;">
<li style="color: #2a9d8f; margin-bottom: 0;">
<b>Basic GroupBy</b>
<p style="color: #264653; font-family: 'Candara', sans-serif; font-size: 14px; margin: 0; line-height:1;">
Group by a single column and calculate the mean of each group.
</p>
<p style="color: #2a9d8f; font-family: 'Candara', sans-serif; font-size: 14px; margin: 0; line-height:1;">
<i>Example:</i> <code>df.groupby('column_name').mean()</code>
</p>
</li>

<li style="color: #2a9d8f; margin-bottom: 0;">
<b>Multiple Columns</b>
<p style="color: #264653; font-family: 'Candara', sans-serif; font-size: 14px; margin: 0; line-height:1;">
Group by multiple columns and calculate the sum of each group.
</p>
<p style="color: #2a9d8f; font-family: 'Candara', sans-serif; font-size: 14px; margin: 0; line-height:1;">
<i>Example:</i> <code>df.groupby(['column1', 'column2']).sum()</code>
</p>
</li>

<li style="color: #2a9d8f; margin-bottom: 0;">
<b>Custom Aggregation</b>
<p style="color: #264653; font-family: 'Candara', sans-serif; font-size: 14px; margin: 0; line-height:1;">
Use custom aggregation functions on a group.
</p>
<p style="color: #2a9d8f; font-family: 'Candara', sans-serif; font-size: 14px; margin: 0; line-height:1;">
<i>Example:</i> <code>df.groupby('column_name').agg({'column_to_aggregate': 'sum'})</code>
</p>
</li>
</ol>

<h3 style="color: #2a9d8f; font-family: 'Candara', sans-serif; margin-top: 0;">Syntax</h3>

<h3 style="color: #2a9d8f; font-family: 'Candara', sans-serif; margin-top: 0;">Parameters</h3>
<ul style="font-family: 'Candara', sans-serif; font-size: 14px; line-height:1;">
<li><b>by:</b> Column(s) to group by. Can be a column name, list of column names, or a function.</li>
<li><b>axis:</b> Axis to group along (default 0 for rows).</li>
<li><b>level:</b> If axis has a MultiIndex, group by a particular level.</li>
<li><b>as_index:</b> Whether to set the group labels as the index (default True).</li>
<li><b>sort:</b> Whether to sort the group labels (default True).</li>
<li><b>group_keys:</b> Whether to include group keys in the result (default True).</li>
<li><b>dropna:</b> Whether to drop groups with missing values (default True).</li>
