```python
import pandas as pd
import matplotlib.pyplot as plt

```

```python
data = {
    'Data_1': [70, 78, 84, 90, 96, 100, 104, 110, 116, 122, 130],
    'Data_2': [40, 56, 68, 80, 92, 100, 108, 120, 132, 144, 160]
}

df = pd.DataFrame(data)
```

```python
mean_1 = df['Data_2'].mean()

Var_2 = df['Data_2'].var()

Var_2
std_2 = df['Data_2'].std()
std_2
```

```python


import seaborn as sns
sns.kdeplot(df['Data_2'], fill=True)
plt.axvline(mean_1, linestyle='--',label=f'Mean = {mean_1:.2f}')

print(Var_2)
```

```python
df['Data_2'].std()
```

```python
# KDE plot
#'Data_2' : [80, 90, 100, 110, 120]
import matplotlib.pyplot as plt

mean = df['Data_2'].mean()
std = df['Data_2'].std()
print(mean-std)
print(62-std)

plt.show()
```

```python
import seaborn as sns
sns.kdeplot(df['Data_2'], fill=True)
plt.axvline(mean, linestyle='--',label=f'Mean = {mean:.2f}')
plt.axvline(mean - std, linestyle=':', label=f'Mean - SD = {mean-std:.2f}')
plt.axvline(mean + std, linestyle=':', label=f'Mean + SD = {mean+std:.2f}')

```
