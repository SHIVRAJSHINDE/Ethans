```python
import pandas as pd
```

```python
data = {
    'Data_1': [70, 78, 84, 90, 96, 100, 104, 110, 116, 122, 130],
    'Data_2': [40, 56, 68, 80, 92, 100, 108, 120, 132, 144, 160]
}

df = pd.DataFrame(data)
```

```python
Data_1_mean = df['Data_1'].mean()
Data_1_mean

Data_1_Var = df['Data_1'].var()
Data_1_Var
```

```python
Data_2_mean = df['Data_2'].mean()
Data_2_mean

Data_2_var = df['Data_2'].var()
Data_2_var

```

```python
import matplotlib.pyplot as plt
import seaborn as sns

DT1_MEAN = df['Data_1'].mean()
print(DT1_MEAN)
DT1_STD = df['Data_1'].std()
print(DT1_STD)

plt.figure(figsize=(5,3))
sns.kdeplot(df['Data_1'])
plt.axvline(DT1_MEAN,linestyle="--")

plt.axvline(DT1_MEAN-DT1_STD,linestyle='--',color='orange')
plt.axvline(DT1_MEAN+DT1_STD,linestyle='--',color='orange')

plt.axvline(DT1_MEAN-(2*DT1_STD),linestyle='--',color='red')
plt.axvline(DT1_MEAN+(2*DT1_STD),linestyle='--',color='red')

plt.axvline(DT1_MEAN-(3*DT1_STD),linestyle='--',color='yellow')
plt.axvline(DT1_MEAN+(3*DT1_STD),linestyle='--',color='yellow')

```
