```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    'Normal': [63, 48, 51, 37, 45,51, 40, 53, 45, 47,45, 67, 50, 40, 57,39, 52, 32, 38, 52],
    '+VeKurtosis': [50, 50, 50, 50, 50,50, 50, 50, 50, 50,50, 50, 50, 50, 49,51, 45, 55, 30, 70],
    '-VeKurtosis': [30, 32, 34, 36, 38,41, 43, 46, 48, 50,50, 52, 54, 57, 59,62, 64, 66, 68, 70]
}

df = pd.DataFrame(data)

```

```python
# Plotting the density distributions
plt.figure(figsize=(8, 5))
for col in df.columns:
    sns.kdeplot(df[col], label=col)

plt.xlabel("Values")
plt.ylabel("Density")
plt.legend()
plt.show()

df
```
