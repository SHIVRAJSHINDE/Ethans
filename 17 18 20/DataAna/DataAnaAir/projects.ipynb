```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(15, 5))

# Create the count plot without passing a palette argument
ax = sns.countplot(x=df["Airline"])

# Add counts on top of each bar
ax.bar_label(ax.containers[0], fmt="%d", padding=3)

plt.xticks(rotation=45)

plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(15, 5))

# Create the count plot with categories sorted by their counts
ax = sns.countplot(x=df["Airline"], order=df["Airline"].value_counts().index)


plt.xticks(rotation=45)

plt.show()
```

```python
# Price Distri bution against Source

Index = df['Source'].value_counts().index

plt.figure(figsize=(10,4))
ax = sns.countplot(x=df['Source'],order=Index)
ax.bar_label(ax.containers[0], fmt="%d", padding=3)

plt.show()
```

```python
# pRICE aNALYSIS
```

```python
A_i = df.groupby('Airline')['Price'].sum().sort_values(ascending=False).index
A_i


plt.figure(figsize=(10,10))
ax = sns.barplot(y=df['Airline'],x=df['Price'],order=A_i,ci=None,estimator=sum)
ax.bar_label(ax.containers[0], fmt="%.0f", padding=3)
plt.xticks(rotation=45)
plt.show()
```

```python
# Sum of PRICE 
S_i= df.groupby('Source')['Price'].sum().sort_values(ascending=False).index

ax = sns.barplot(x=df['Source'],y=df['Price'],estimator=sum,order=S_i)
# Add counts on top of each bar
ax.bar_label(ax.containers[0], fmt="%d", padding=3)


plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Assuming S_i is already defined:
S_i = df.groupby('Source')['Price'].max().sort_values(ascending=False).index

plt.figure(figsize=(10, 6))

# Create the violin plot with the sorted order
sns.violinplot(x=df["Source"],y=df["Price"],order=S_i,inner="quartile")

plt.xticks(rotation=45)
plt.title("Price Distribution by Source", fontsize=14, weight="bold")
plt.xlabel("Source", fontsize=12)
plt.ylabel("Price (INR)", fontsize=12)

plt.show()
```

```python

source = df['Source'].unique()


plt.figure(figsize=(15,8))

for k,v in enumerate(source):
    data = df[df['Source']==v]
    plt.subplot(1,5,k+1)
    sns.violinplot(data=data,y='Price')
    plt.xlabel(v)
plt.show()
```
