# Source,Destination,Route,Total_Stops,Price


```python
df[df['Price'].isnull()]
Missing_Price = df['Price'].mean()
Missing_Price
df.fillna({'Price':Missing_Price},inplace=True)
df[df['Price'].isnull()]
```

```python
df[df['Total_Stops'].isnull()]
Missing_Total_Stops = df['Total_Stops'].mode()[0]
Missing_Total_Stops
df.fillna({'Total_Stops':Missing_Total_Stops},inplace=True)
df[df['Total_Stops'].isnull()]
```

```python
df[df['Route'].isnull()]
Missing_Route = df['Route'].mode()[0]
df.fillna({'Route':Missing_Route},inplace=True)
df[df['Route'].isnull()]
```

```python
df[df['Destination'].isnull()]
Missing_Dest = df['Destination'].mode()[0]
df.fillna({'Destination':Missing_Dest},inplace=True)
df[df['Destination'].isnull()]
```

```python
df[df['Source'].isnull()]
Missing_Source = df['Source'].mode()[0]
Missing_Source
df.fillna({'Source':Missing_Source},inplace=True)
df[df['Source'].isnull()]

```
