```python
from sklearn.preprocessing import OrdinalEncoder, OneHotEncoder, StandardScaler

Ordinal_Encoding = ('Ordinal_Encoding',OrdinalEncoder(categories=[Education_Seq, Employment_Stability_Seq]),Ordinal_Col,)
OneHot_Encoding =  ('OneHot_Encoding',OneHotEncoder(sparse_output=False,drop="first", handle_unknown="ignore"),Onehot_Col,)
Scaling = ("Scaling", StandardScaler(), Numeric_Col)



```

```python
from sklearn.compose import ColumnTransformer

trans_Obj = ColumnTransformer(
        transformers=[Scaling]).set_output(transform="pandas")

```
