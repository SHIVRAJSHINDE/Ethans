```python
import pandas as pd
```

```python
# Get Training DATA
X_Train = pd.read_excel("Data_Train_Test_Split\X_Train.xlsx")

y_Train_true = pd.read_excel("Data_Train_Test_Split\y_Train_true.xlsx")

X_Train_Trans = pd.read_excel("Data_Transformed\X_Train_Trans.xlsx")


```

```python
# Get Testing DATA 
X_Test = pd.read_excel("Data_Train_Test_Split\X_Test.xlsx")
y_Test_true = pd.read_excel("Data_Train_Test_Split\y_Test_true.xlsx")


X_Test_Trans = pd.read_excel("Data_Transformed\X_Test_Trans.xlsx")

```

```python
# Linear Regression
import joblib
from sklearn.linear_model import LinearRegression

LinearR_Model = LinearRegression()

LinearR_Model.fit(X_Train_Trans,y_Train_true)

joblib.dump(LinearR_Model,"Models\Linear_Model.pkl")
```

```python
Linear_Model_Obj = joblib.load("Models\Linear_Model.pkl")
```

```python
# Generate y_train_pred
y_train_pred = Linear_Model_Obj.predict(X_Train_Trans)

y_train_pred

```

```python
y_test_pred = Linear_Model_Obj.predict(X_Test_Trans)
y_test_pred

```

```python
#Evaluation Phase
from sklearn.metrics import mean_squared_error, mean_absolute_error,r2_score
import numpy as np

def linear_Metrics(x_Data,True_Value,Pred_Value):

    MSE = mean_squared_error(True_Value,Pred_Value) 
    MAE = mean_absolute_error(True_Value,Pred_Value)
    RMSE = np.sqrt(MSE)
    R2 = r2_score(True_Value,Pred_Value)

    n = len(x_Data)
    p = x_Data.shape[1]
    A_R2 = 1-(1-R2)*(n-1)/(n-p-1)

    merticsDict = {"MSE": MSE,
                   "MAE":MAE,
                   "RMSE": RMSE,
                   "R2": R2,
                   "A_R2":A_R2}
    return merticsDict


```

```python
linear_Metrics(X_Train,y_train_true,y_train_pred)
```

```python
linear_Metrics(X_Test,y_Test_true,y_test_pred)
```
