```python
import pandas as pd
```

```python
X_test = pd.read_excel("Data_Train_Test\X_test.xlsx")
X_train = pd.read_excel("Data_Train_Test\X_train.xlsx")
y_test_true = pd.read_excel("Data_Train_Test\y_test_true.xlsx")
y_train_true = pd.read_excel("Data_Train_Test\y_train_true.xlsx")
```

```python
# Read Transform
X_test_trans = pd.read_excel("Data_Trans\X_test_trans.xlsx")
X_train_trans = pd.read_excel("Data_Trans\X_train_trans.xlsx")
```

```python
# Start Training
# Linear Regression

from sklearn.linear_model import LinearRegression

LinearRegressionOBJ = LinearRegression()

LinearRegressionOBJ.fit(X_train_trans,y_train_true)


```

```python
# Save Linear Model
import joblib 
joblib.dump(LinearRegressionOBJ,"Model_Files/LinearModel.pkl")
```

```python
LinearModel_1 = joblib.load("Model_Files\LinearModel.pkl")
```

```python
# PREDICT y_train_pred
# PREDICT y_test_pred
y_train_pred = LinearModel_1.predict(X_train_trans)
y_train_pred

```

```python
y_test_pred = LinearModel_1.predict(X_test_trans)
```

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error,r2_score
import numpy as np

def linear_Metrics(x_Data,True_Value,Pred_Value):

    MSE = mean_squared_error(True_Value,Pred_Value) 
    MAE = mean_absolute_error(True_Value,Pred_Value)
    RMSE = np.sqrt(MSE)
    R2 = r2_score(True_Value,Pred_Value)

    n = len(x_Data)
    p = x_Data.shape[1]
    A_R2 = 1-(1-R2)*(n-1)/(n-p-1)

    merticsDict = {"MSE": MSE,
                   "MAE":MAE,
                   "RMSE": RMSE,
                   "R2": R2,
                   "A_R2":A_R2}
    return merticsDict


```

```python
# Training Evaluation
linear_Metrics(X_train,y_train_true,y_train_pred)
```

```python
# Testing Evaluation
linear_Metrics(X_test,y_test_true,y_test_pred)
```
