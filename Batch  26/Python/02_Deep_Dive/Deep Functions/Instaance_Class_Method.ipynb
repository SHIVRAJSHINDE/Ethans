# Instance Method

```python
class Dog():
    def __init__(self):
        self.dog_Age=14 # Instance Varialbe

dog_1 = Dog()
dog_1.dog_Age
```

```python
class Dog():
    def __init__(self,dog_Age):
        self.dog_Age=dog_Age # Instance Varialbe

    def abc(self):              # Instance Method
        print(self.dog_Age)

    def my_dog_details(self,colour):
        print(f"my Dog has {colour}color")

dog_1 = Dog(14)
dog_1.abc()
dog_1.my_dog_details("Brown")
```

# Class Variable

```python
class School:
    school_name = "Bharatiya Vidhya Bhavan"     # Class Variable
School.school_name
```

```python
s_obj_01 = School()
s_obj_01.school_name = "GreenWood School "
s_obj_01.school_name
```

```python
class Bankig:
    Int_Rate = 0.08     # 8%Class Variable

    @classmethod
    def change_int_Rage(cls):
        cls.Int_Rate = cls.Int_Rate + (cls.Int_Rate*0.02)
        print(cls.Int_Rate)

Bankig.Int_Rate
Bankig.change_int_Rage()
```
