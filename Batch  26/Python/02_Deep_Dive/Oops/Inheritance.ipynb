```python
class ParentClass:
    def Parent_Class_method(self):
        return "Parent Class method I have 50 Acres of Land"

class ChildClass(ParentClass):
    def ChildClass_method(self):
        return "Child Class Method I have Bike"

Parent_Obj_01 = ParentClass()
Parent_Obj_01.Parent_Class_method()
```

```python
Child_Obj_01 =ChildClass()
print(Child_Obj_01.ChildClass_method())
Child_Obj_01.Parent_Class_method()
```

# Multi Level Inheritance

```python
class GrandParentsClass:
    def GrandParents_Method(self):
        return "GP_Method I have 500 Acres of Land"

class ParentClass(GrandParentsClass):
    def Parent_Class_method(self):
        return "Parent Class method I have 50 Acres of Land"

class ChildClass(ParentClass):
    def ChildClass_method(self):
        return "Child Class Method I have Bike"
GP_Obj_01 = GrandParentsClass()
GP_Obj_01.GrandParents_Method()
```

```python
Parent_Obj_01 = ParentClass()
print(Parent_Obj_01.Parent_Class_method())
print(Parent_Obj_01.GrandParents_Method())
```

```python
Child_Obj_01 = ChildClass()
print(Child_Obj_01.ChildClass_method())
print(Child_Obj_01.Parent_Class_method())
print(Child_Obj_01.GrandParents_Method())
```

# Multiple Inheritance

```python
class Mother:
    def Mother_Method(self):
        return "I have Gold"
class Father:
    def Father_method(self):
        return "I have House and Money"

class child(Mother,Father):
    def child_method(self):
        return "I have Books"



chil_Obj_01 = child()
chil_Obj_01.Father_method()
chil_Obj_01.Mother_Method()
```
