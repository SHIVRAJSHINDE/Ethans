# Method OVerloading

```python
# Simple method without overloading
class myClass:
    def sum_4(self,a,b,c,d):
        return a+b+c+d
    def sum_3(self,a,b,c):
        return a+b+c
    def sum_2(self,a,b):
        return a+b
    
my_Obj = myClass()
my_Obj.sum_3(1,2,3)
my_Obj.sum_2(1,1)
my_Obj.sum_4(1,2,3,4)
```

```python
#  With overloading
class myClass:
    def sum_Overlaoded(self,a=None,b=None,c=None,d=None):
        if a != None and b !=None and c !=None and d !=None:
            return a+b+c+d
        if a != None and b !=None and c !=None:
            return a+b+c
        elif a != None and b !=None:
            return a+b
        else:
            return "Pls pass atleast Two Values for sum"
my_Obj = myClass()
my_Obj.sum_Overlaoded(1,2,3,4)
```

```python
# Simple method without overloading
class myClass_01:
    def sum(self,a,b,c,d):
        return a+b+c+d
    def sum(self,a,b,c):
        return a+b+c
    def sum(self,a,b):
        return a+b
    
my_Obj = myClass_01()
my_Obj.sum(1,2,3,4)
```

# Method Overridding

```python
class Animal:
    def sound(self):
        return "Animal makes Sound"
class Dog(Animal):    
    def sound(self):
        return "Barks"
class Cat(Animal):    
    def sound(self):
        return "Meow"

Dog_Obj = Dog()
Dog_Obj.sound()
```

```python
class parentClass:
    def sound(self):
        return "ParentClass Animal makes Sound"
class ChildClass(parentClass):    
    def sound(self):
        return "ChiLD CLASS Barks"


Dog_Obj = ChildClass()
Dog_Obj.sound()
```

```python
class P_C_Interest:
    def Interest(self):
        return 0.011

class c_c_Home_Interese(P_C_Interest):
    def Interest(self):
        return 0.08
Home_oBJ = c_c_Home_Interese()
Home_oBJ.Interest()
```

```python
# Uuse of Super Keyword
class P_C_Interest:
    def Interest(self):
        print(0.011) 

class c_c_Home_Interese(P_C_Interest):
    def Interest(self):
        super().Interest()
        print(0.08) 
Home_oBJ = c_c_Home_Interese()
Home_oBJ.Interest()
```
