```python
import pandas as pd
```

```python
df = pd.read_excel("Data_Raw\Classification_LoanSanction.xlsx")
df.head(2)
```

```python
df = df.drop(['ID','Credit_Risk'],axis=1)
df
```

```python
# Split data in to X and y

X = df.drop('Loan_Sanctioned',axis=1)
X
y = df['Loan_Sanctioned']

y

X
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train_true, y_test_true = train_test_split(X,y,test_size=0.20,random_state=42)

```

```python

X_train.to_excel("Data_Train_Test/X_train.xlsx",index=False)
X_test.to_excel("Data_Train_Test/X_test.xlsx",index=False)
y_train_true.to_excel("Data_Train_Test/y_train_true.xlsx",index=False)
y_test_true.to_excel("Data_Train_Test/y_test_true.xlsx",index=False)
```

```python
X_test.head(2)
```

```python
#Numerical Columns
Numeric_Col = ['Age','Income','Credit_Score','Loan_Amount','Employment_Years','Existing_Loans']
#Categorical
#   Ordinal Columns
Ordinal_Col = ['Education','Employment_Stability']

Education_Seq = ['High School', 'Graduate', 'Postgraduate']
Employment_Stability_Seq = ['Low', 'Medium', 'High']

#Onehot
#Onehot_Col = ['Employment_Type']
Onehot_Col = ['Gender','Employment_Type','Marital_Status','Property_Ownership']
```

```python
from sklearn.preprocessing import OrdinalEncoder, OneHotEncoder, StandardScaler

Ordinal_Encoding = ('Ordinal_Encoding',OrdinalEncoder(categories=[Education_Seq, Employment_Stability_Seq]),Ordinal_Col,)
OneHot_Encoding =  ('OneHot_Encoding',OneHotEncoder(sparse_output=False,drop="first", handle_unknown="ignore"),Onehot_Col,)
Scaling = ("Scaling", StandardScaler(), Numeric_Col)

```

```python
from sklearn.compose import ColumnTransformer

trans_Obj = ColumnTransformer(
        transformers=[Ordinal_Encoding,OneHot_Encoding,Scaling]).set_output(transform="pandas")

```

```python
X_train_trans = trans_Obj.fit_transform(X_train)

X_train_trans.to_excel("Data_Trans/X_train_trans.xlsx",index=False)
```

```python
# Save Transform.PKL 
import joblib
joblib.dump(trans_Obj,"Trans_File/transform.pkl")
```
