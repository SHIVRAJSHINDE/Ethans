```python
import pandas as pd
import joblib

X_test = pd.read_csv("Airtifacts\Clnd_Data\X_test.csv")

X_train = pd.read_csv("Airtifacts\Clnd_Data\X_train.csv")
X_train_trans = pd.read_csv("Airtifacts\Trans_Data\X_train_trans.csv")
X_test_trans = pd.read_csv("Airtifacts\Trans_Data\X_test_trans.csv")
y_train_true = pd.read_csv("Airtifacts\Clnd_Data\y_train_true.csv")
y_test_true = pd.read_csv("Airtifacts\Clnd_Data\y_test_true.csv")
```

```python
from sklearn.linear_model import LinearRegression
LinerModel = LinearRegression()
LinerModel.fit(X_train_trans,y_train_true)
joblib.dump(LinerModel,"Airtifacts/pklModels/LinerModel.pkl")
```

```python
# Load the Model
LinerModel_01 = joblib.load("Airtifacts/pklModels/LinerModel.pkl")
```

```python
# Extract y_train_pred
y_train_pred = LinerModel_01.predict(X_train_trans)
y_train_pred
```

```python
# Extract y_test_pred
y_test_pred = LinerModel_01.predict(X_test_trans)
y_test_pred
```

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, root_mean_squared_error,r2_score

def linear_Metrics(X_Data,y_true,y_pred):
    MSE = mean_squared_error(y_true,y_pred) 
    MAE = mean_absolute_error(y_true,y_pred)
    RMSE = root_mean_squared_error(y_true,y_pred)
    R2 = r2_score(y_true,y_pred)

    n = len(X_Data)
    p = X_Data.shape[1]
    A_R2 = 1-(1-R2)*(n-1)/(n-p-1)

    print(MSE)
    print(MAE)
    print(RMSE)
    print(R2)
    print(A_R2)
```

```python
# Training Evaluation
# y_train_true
# y_test_pred

linear_Metrics(X_train,y_train_true,y_train_pred)
```

```python
# Tresting Evaluation
# y_test_true
# y_test_pred

linear_Metrics(X_test,y_test_true,y_test_pred)
```
