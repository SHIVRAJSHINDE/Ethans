```python
import pandas as pd
import joblib

X_test = pd.read_csv("Airtifacts\Clnd_Data\X_test.csv")
X_train = pd.read_csv("Airtifacts\Clnd_Data\X_train.csv")

X_train_trans = pd.read_csv("Airtifacts\Trans_Data\X_train_trans.csv")
X_test_trans = pd.read_csv("Airtifacts\Trans_Data\X_test_trans.csv")
y_train_true = pd.read_csv("Airtifacts\Clnd_Data\y_train_true.csv")
y_test_true = pd.read_csv("Airtifacts\Clnd_Data\y_test_true.csv")
```

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, root_mean_squared_error,r2_score

def linear_Metrics(X_Data,y_true,y_pred):
    MSE = mean_squared_error(y_true,y_pred) 
    MAE = mean_absolute_error(y_true,y_pred)
    RMSE = root_mean_squared_error(y_true,y_pred)
    R2 = r2_score(y_true,y_pred)

    n = len(X_Data)
    p = X_Data.shape[1]
    A_R2 = 1-(1-R2)*(n-1)/(n-p-1)

    print(MSE)
    print(MAE)
    print(RMSE)
    print(R2)
    print(A_R2)
```

```python
from sklearn.linear_model import LinearRegression,Lasso,Ridge,ElasticNet
from sklearn.model_selection import GridSearchCV

models = {
    "Linear_Reg": LinearRegression(),
    "Ridge_Reg": Ridge(),
    "Lasso_Reg": Lasso(),
    "Elastic_Net": ElasticNet()}

params = {
    "Linear_Reg": {},
    "Lasso_Reg": {"alpha": [0.01, 0.1, 1.0]},
    "Ridge_Reg": {"alpha": [0.1, 1.0, 10.0]},
    "Elastic_Net": {"alpha": [0.01, 0.1, 1.0]},
}
def Model_Training(models,params,X_train_trans,X_test_trans,X_train,X_test):
    for model_Name, act_Model in models.items():
        grid_obj = GridSearchCV(act_Model,params[model_Name])
        grid_obj.fit(X_train_trans,y_train_true)    # Training phase
        best_model_with_par = grid_obj.best_estimator_
        joblib.dump(best_model_with_par,f"Airtifacts/pklModels/{model_Name}.pkl")
        y_train_pred = best_model_with_par.predict(X_train_trans)
        print("Train")
        linear_Metrics(X_train,y_train_true,y_train_pred)
        print("Test")
        y_test_pred = best_model_with_par.predict(X_test_trans)
        linear_Metrics(X_test,y_test_true,y_test_pred)
Model_Training(models,params,X_train_trans,X_test_trans,X_train,X_test)

```
