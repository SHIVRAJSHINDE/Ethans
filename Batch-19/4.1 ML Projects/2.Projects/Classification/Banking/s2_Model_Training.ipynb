# Load the DATA

```python
import pandas as pd
```

```python
# Load the Train and Test data
X_train = pd.read_excel("Data_Train_Test\X_train.xlsx")
X_test = pd.read_excel("Data_Train_Test\X_test.xlsx")
y_train_true = pd.read_excel("Data_Train_Test\y_train_true.xlsx")
y_test_true = pd.read_excel("Data_Train_Test\y_test_true.xlsx")
```

```python
# Read the Transform Data
X_train_trans = pd.read_excel("Data_Trans\X_train_trans.xlsx")
X_test_trans = pd.read_excel("Data_Trans\X_test_trans.xlsx")
```

# Linear Regression

```python
# Load the Model
from sklearn.linear_model import LogisticRegression

Log_Reg_Obj = LogisticRegression()
Log_Reg_Obj
```

```python
# Train the model using X_train_trans and y_train_true

Log_Reg_Obj.fit(X_train_trans,y_train_true)
```

```python
import joblib
joblib.dump(Log_Reg_Obj,"Model_Files/Log_Reg_Obj.pkl")
```

```python
# Get Predicted Data
# Training predicted Data
import joblib 

Log_Reg_Obj_01 = joblib.load("Model_Files\Log_Reg_Obj.pkl")

```

```python
# Generate y_train_pred 
y_train_pred = Log_Reg_Obj_01.predict(X_train_trans)
y_train_pred
```

```python
# Generate y_test_pred


y_test_pred = Log_Reg_Obj_01.predict(X_test_trans)
y_test_pred

```

# Evaluation Functoin for Train Test Data


```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

def ClassificationMetrics(y_true,y_pred):
    Acc = accuracy_score(y_train_true,y_train_pred) 
    prec = precision_score(y_train_true,y_train_pred)
    reca = recall_score(y_train_true,y_train_pred)
    f1 = f1_score(y_train_true,y_train_pred)
    newDict = {'Acc': Acc,  
               'prec':prec , 
               'reca':reca , 
               'f1': f1 }

    return newDict
```

```python
# Train Evaluation
ClassificationMetrics(y_train_true,y_train_pred)

```

```python
# Testing Evalution 
ClassificationMetrics(y_test_true,y_test_pred)

```
