```python
import pandas as pd
```

```python
df = pd.read_excel("Data_Raw\CorporateSalary.xlsx")
df.head(2)
```

```python
y = df['Salary']
y
```

```python
X = df.drop('Salary',axis=1)
X
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train_true, y_test_true = train_test_split(X,y,test_size=0.20,random_state=42)

```

```python
# Save Training
X_train.to_excel("Data_Train_Test/X_train.xlsx",index=False)
y_train_true.to_excel("Data_Train_Test/y_train_true.xlsx",index=False)

```

```python
# Save Testing
X_test.to_excel("Data_Train_Test/X_test.xlsx",index=False)
y_test_true.to_excel("Data_Train_Test/y_test_true.xlsx",index=False)
```

```python
# Transformation 
Numerica_Colums = ['TenureYears','ProjectsCompleted','PerformanceScore']
Ordered_Columns = ['Designation', 'Education']
One_Hot_Columns = ['Department']

Designation_Order = ['Junior Analyst', 'Senior Analyst', 'Lead Consultant', 'Manager', 'Senior Manager','Director']
Education_Order = ["Bachelor's", "Master's", 'PhD']


# Transformation Actual Process
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder, StandardScaler


OrdinalEncoding_01 = ("Oridnal_Encoder",OrdinalEncoder(categories=[Designation_Order, Education_Order]),Ordered_Columns,)
OneHotEncoding_01 =  ("One_Hot_Encoder",OneHotEncoder(sparse_output=False, drop="first", handle_unknown="ignore"),One_Hot_Columns,)
Scaling = ("Numerical_Coluns", StandardScaler(), Numerica_Colums)

transformer_engine = ColumnTransformer(transformers=[OrdinalEncoding_01,OneHotEncoding_01,Scaling]).set_output(transform="pandas")

transformer_engine

```

```python
X_train_trans = transformer_engine.fit_transform(X_train)
X_train_trans
```

```python
import joblib

joblib.dump(transformer_engine,"Trans_File/TransFrom.pkl")
```

```python
TransFromFile = joblib.load("Trans_File/TransFrom.pkl")

```

```python
# Transform X_Test

X_test_trans = TransFromFile.fit_transform(X_test)
X_test_trans

```

```python
X_train_trans.to_excel("Data_Trans/X_train_trans.xlsx")
X_test_trans.to_excel("Data_Trans/X_test_trans.xlsx")
```
