```python
import pandas as pd

file_path = r"D:\Old Projects\MLProjects\ZDatasets\DummyData\Data_Train_Airline.xlsx"

# Read the Excel file
df = pd.read_excel(file_path)

# Display first few rowsabs
df.head()
```

```python
# Import matplotlib
import matplotlib.pyplot as plt
# Documentation of the matplotlib https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.bar.html
# matplotlib.pyplot.bar(x, height, width=0.8, bottom=None, *, align='center', data=None, **kwargs)

plt.figure(figsize=(10,5))
f = 15
plt.bar(df['Airline'],df['Price'],width=0.8,align='center',color=['r'])
plt.xticks(rotation=45)
plt.xlabel("Airline",fontsize=f)
plt.ylabel("Price",fontsize=f)
plt.title('Airline Vs Price',fontsize=f)
plt.show()

```

```python
newDf = df.groupby(['Airline'])['Price'].sum()
newDf = pd.DataFrame(newDf)
newDf = newDf.sort_values(['Price'],ascending=False)
newDf = newDf.reset_index()
print(newDf)

sns.barplot(x=newDf['Airline'],y=newDf['Price'],errorbar=None)
plt.xticks(rotation=90)
plt.show()

```

```python
newDf = df.groupby(['Airline'])['Price'].mean()
newDf = pd.DataFrame(newDf)
newDf = newDf.sort_values(['Price'],ascending=False)
newDf = newDf.reset_index()
print(newDf)

sns.barplot(x=newDf['Airline'],y=newDf['Price'],errorbar=None)
plt.xticks(rotation=90)
plt.title('Comparision of Max price of All Airlines')
plt.show()


# Homework/Task
# mean max sum min of categories like Airline Source 

# Airline and Mean
# Airline and max
# Airline and sum
# Airline and min

# Source and mean
# Source and max
# Source and sum
# Source and min

```
