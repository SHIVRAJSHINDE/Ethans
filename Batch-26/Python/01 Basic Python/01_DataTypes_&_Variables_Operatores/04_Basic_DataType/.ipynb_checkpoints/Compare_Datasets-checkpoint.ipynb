```python
abcList1.extend(abcList2)
abcList1
```

```python
abcTup1 = ("cherry", "orange")
abcTup2 = ("apple", "banana")
neT = abcTup1+abcTup2
neT
```

```python
abcSet1 = {"cherry", "orange"}
abcSet2 = {"apple", "banana"}

```

```python
abc = abcSet1.union(abcSet2)
abc
```

```python
abcSet1.update(abcSet2)
abcSet1
```
