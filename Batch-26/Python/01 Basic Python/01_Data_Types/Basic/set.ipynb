```python
# Add Items
mySet = {"apple", "banana", "mango"}

mySet.add("PineApple")
mySet
```

```python
mySet = {"apple", "banana", "mango"}

abc = ["Pipeapple","Grapes"]
mySet.update(abc)
mySet
```

```python
new_set = {'Grapes', 'Pipeapple', 'apple', 'banana', 'mango'}
new_set.remove("banana")
new_set
```

```python
new_set = {'Grapes', 'Pipeapple', 'apple', 'banana', 'mango'}
new_set.discard("banana")
new_set
```

```python
new_set = {'Grapes', 'Pipeapple', 'apple', 'banana', 'mango'}
new_set.pop()
new_set
```

```python
new_set = {'Grapes', 'Pipeapple', 'apple', 'banana', 'mango'}
new_set.clear()
new_set
```
