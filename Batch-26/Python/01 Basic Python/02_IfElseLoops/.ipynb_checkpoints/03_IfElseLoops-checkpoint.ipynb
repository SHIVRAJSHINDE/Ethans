# If Condition

```python
# IF condition
# check Value
# please check if the value is less than or equal to 10


i = 12

if i <5:
    print("Given Values is less that 5")
elif i<8:
    print("Given Values is less that 8")
elif i<12:
    print("Given Values is less that 12")
elif i==12:
    print("Given Values equal to 12")
else:
    print("Given Values is greater than 12")

```

```python
# check if the the given age falls in to infant or Adult or Old Person 
# infant <=2 
# child <=15 or 
# Adult <=55
# old person(greater than 55)

age = 55

if age<=2:
    print("infant")
elif age<=15:
    print("child")
elif age<55:
    print("Adult")
else:
    print("Old Person")
```


# While Loop




```python
# print number from 0 to 10
i = 0

while i<=10:
    print(i)
    i = i + 1
```

```python
# print only even numbers from 0 to 10 
i = 0

while i<=10:
    if i%2==0:
        print(i)
    
    i = i + 1


```

```python
For loop work on below
# range
# Iterable Data types
    # list
    # tupple
    # Set
    # string
    # Dict
```

```python
# range
for i in range(0,10):
    print(i)
```

```python
# Set DATATYPE
setOne = {1,2,'q','v','b'}

for i in setOne:
    print(i)
```

```python
# String
name = "Rajesh"

for i in name:
    print(i)

```

```python
newDict = {'name':"RAJESH","AGE":32}

for i in newDict.items():
    print(i)
    
```

```python
'''For Loops'''
fruitsList = ["apple", "banana", "cherry","kiwi"]

for i in fruitsList:
    print(i)
```

```python
# Create Functions
'''
inf <=2
child<=12
adult>12
'''

Age = 56

def classifyAge(age):

    if age <=2:
        print("INF")
    elif age<=12:
        print("child")
    elif age<=55:
        print("Adult")
    else:
        print("Old Persion")

classifyAge(age=Age)
```
