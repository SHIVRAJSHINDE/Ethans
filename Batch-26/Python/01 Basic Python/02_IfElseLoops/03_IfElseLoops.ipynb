```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
sns.set(style="whitegrid")

# Beam length
L = 10

# Point loads (kN) and positions (m)
loads = [10, 15, 8]
positions = [2, 5, 8]

# ---- Calculate reactions ----
# Sum of loads
total_load = sum(loads)
print("total_load: ", total_load)

# Take moment about A
moment_A = 0

for load, position in zip(loads, positions):
    moment_A = moment_A + (load * position)
print("moment_A: " , moment_A)

RB = moment_A / L
RA = total_load - RB

print("Reaction at B =", RB, "kN")
print("Reaction at A =", RA, "kN")


# ---- Create beam points ----
x = np.linspace(0, L, 500)

shear = []
moment = []

for xi in x:
    V = RA
    M = RA * xi

    for P, a in zip(loads, positions):
        if xi >= a:
            V = V - P
            M = M - (P * (xi - a))
    shear.append(V)
    moment.append(M)

print("shear: ", shear)
print("moment: ", moment)



```

```python
# ---- Plotting ----
plt.figure(figsize=(12,6))
# Shear Force Diagram
plt.subplot(2,1,1)
plt.plot(x, shear, color='blue', linewidth=2)
plt.fill_between(x, shear, alpha=0.3)
plt.title("Shear Force Diagram")
plt.ylabel("Shear Force (kN)")
plt.xlabel("Distance")

# Bending Moment Diagram
# plt.subplot(row,colum,plotNumber)
plt.subplot(2,1,2)
plt.plot(x, moment, color='red', linewidth=2)
plt.fill_between(x, moment, alpha=0.3)
plt.title("Bending Moment Diagram")
plt.xlabel("Beam Length (m)")
plt.ylabel("Bending Moment (kN-m)")

plt.tight_layout()
plt.show()
```
