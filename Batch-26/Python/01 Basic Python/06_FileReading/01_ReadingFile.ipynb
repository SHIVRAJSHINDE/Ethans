#  Create a File

```python
# Readin is a in build feature of python

filePath = "D:/neData/abc.txt"
file  = open(filePath,"w")
```


# Writting in to file

```python
file.write("this is text file")
file.write("this is dummy file on new Line")
file.close()
```

# Reading File


```python
file  = open(filePath,"r")
abc = file.read()

print(abc)
file.close()
```

# Appending in to File

```python
filePath = "D:/neData/abc.txt"
file = open(filePath,'a')

file.write("--------------------------------------------------------------\n")
file.write("Hello, this is some text written to a text file using Python!\n")
file.close()
```

# Use of WITH KeyWord

```python
# Reading from a file
filePath = "D:/neData/xyz.txt"

with open(filePath,'w') as file:
    file.write("asdfasdfasdf")
    file.write("asdfasdfasdf")
    file.write("asdfasdfasdf")
    file.write("asdfasdfasdf")
    
```
