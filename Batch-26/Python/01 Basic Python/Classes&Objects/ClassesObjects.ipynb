```python
class Vehicle:

    def __init__ (self,make, model):
        self.make = make
        self.model = model

    def get_info(self):
        print("Info of the car is: ",self.make,self.model)

    def drive(self):
        print(self.model,"Can run up to the Speed of 140 km/hr")

        
car = Vehicle("TATA","Safari")
car.drive()

```
