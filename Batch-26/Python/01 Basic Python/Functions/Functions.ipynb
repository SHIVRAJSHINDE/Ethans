```python
# With Arguments
def printSquares(list_01):
    NewDict_12= ""
    NewDict_12 = {}
    for i in list_01:
        NewDict_12[i] = i*i
    print(NewDict_12)
```

```python
list_02 =  [1,2,3]
printSquares(list_02)
```

```python
list_03 = [10,20,30]
printSquares(list_03)
```

```python
list09 = [12,2,10,11,5,3,1,-1,-10]

def addValueLessThan5(list09):
    sortedlist = []
    for i in list09:
        if i < 5:
            sortedlist.append(i)

    print(sortedlist)

addValueLessThan5(list09)
```

```python


list1 = [12,2,10,11,5,3]

def printValueLessthan_5(list1):
    list2 = [x for x in list1 if x < 5]
    print(list2)

printValueLessthan_5(list1)
```
