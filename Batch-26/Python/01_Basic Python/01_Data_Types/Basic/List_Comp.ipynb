```python
# [exp Loop]
# [exp for i in iterable]
[i for i in range(1,10)]
```

```python
# [exp Loop if_condition]
# [exp for i in iterable if_condition]
# Print Even Numbers 
[i for i in range(1,10) if i%2==0]
```

```python
# Print Square of Even Numbers 
[i**2 for i in range(1,10) if i%2==0]
```

```python
newList = []
for i in range(15):
    if i<5:
        newList.append(i*2)
    elif i<10:
        newList.append(i/2)
    else:
        newList.append(i+2)
newList
```

```python
[i*2 if i<5 else i/2 if i<10 else i+2 for i in range(15)]
```

```python
# convert below in dict comprehension
[i*2 if i<5 else i/2 if i<10 else i+2 for i in range(15)]
```

```python
# {Exp_1: Exp_2 Loop}

{i: i*2 if i<5 else i/2 if i<10 else i+2 for i in range(0,15)}
```

```python
{i : (i*2 if i<5 else i/2 if i<10 else i+2) for i in range(15)}
```

```python
{(i*2 if i<5 else i/2 if i<10 else i+2) : i  for i in range(15)}
```
