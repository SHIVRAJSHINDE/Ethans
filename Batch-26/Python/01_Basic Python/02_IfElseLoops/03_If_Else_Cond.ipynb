```python
i = 3

if i <5:
    print("Value is less than 5")
elif i < 7:
    print("Value is less than 7")
elif i < 10:
    print("Value is less than 10")
else:
    print("-----------")
```

```python
Stall_1 = 'Apple'
Stall_1_Price = 160

Stall_2 = 'Mango'
Stall_2_Price = 150

if Stall_1 == "Apple" and Stall_1_Price <= 150:
    print("Buy Apple from Stall_1")
elif Stall_2 == "Apple" and Stall_2_Price <=150:
    print("Buy Apple from Stall_2")
else:
    print("Did not get Apple in Price Range")
```

```python
stall_1 = ["Apple", "Banana", "Cherry", "Grape", "Mango"]
stall_1_price= 150

stall_2 = ["Banana", "Grape", "Mango", "Orange", "Pineapple"]
stall_2_price = 200

if "Apple" in stall_1 and stall_1_price <= 150:
    print("buy Apple from stall 1")
elif "Apple" in stall_2 and stall_2_price <= 150:
    print("buy Apple from stall 2")
else:
    print("Apple is not available in either stall")
```

```python
# IF condition
# check Value
# please check if the value is less than or equal to 10


i = 12

if i <5:
    print("Given Values is less that 5")
elif i<8:
    print("Given Values is less that 8")
elif i<12:
    print("Given Values is less that 12")
elif i==12:
    print("Given Values equal to 12")
else:
    print("Given Values is greater than 12")

```

```python
# check if the the given age falls in to infant or Adult or Old Person 
# infant <=2 
# child <=15 or 
# Adult <=55
# old person(greater than 55)

age = 55

if age<=2:
    print("infant")
elif age<=15:
    print("child")
elif age<55:
    print("Adult")
else:
    print("Old Person")
```
