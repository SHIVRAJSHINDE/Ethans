```python
list01 = [1,2,3,4]
list02 = []

def myFunc5(func):
    def wrapper(l1):
        l1 = [int(i) if isinstance(i, str) else i for i in l1]
        l2 = func(l1)
        d1 = {}
        for i in range(len(l2)):
            d1[l1[i]] = l2[i]
        print(d1)
    return wrapper

@myFunc5
def originalFunc(l1):
    for i in l1:
        list02.append(i**2)
    return list02
        
originalFunc(list01)

```
