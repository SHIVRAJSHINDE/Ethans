<p style="color: #FFFFFF; font-family: 'Candara', sans-serif; font-size: 18px; line-height: 1.5; margin-top: 15px;">

# <h2 style="color: #2a9d8f; font-family: 'Candara', sans-serif; font-weight: bold; text-align: left;margin:-60px 0 0 30px; padding-bottom: 10px; border-bottom: 2px solid #264653;">.fillna </h2>

<ul style="margin: 0; padding-left: 15px; list-style-type: none;">
<li style="margin-bottom: 0;"><code>df.fillna(0)</code> — Replace all NaNs with 0.</li>

<li style="margin-bottom: 0;"><code>df.fillna({'col1':0,'col2':1})</code> — Fill NaNs in 'col1' and 'col2' with 0 and 1.</li>

Adding new columns to a DataFrame can be done in several ways. Below are the most popular methods:

<ul style="font-family:Candara, sans-serif; fon6-size:17px; line-height:1; padding-left:15px; margin:0; list-style-type:none;">

<p style="margin:0 0 0 20px;">Description: You can directly add a new column by assigning values to a new column label.</p>
<p style="margin:0 0 0 20px;">Syntax: <code>df['new_column_name'] = value</code></p>

<li style="color:#2a9d8f; margin:0;">

<li style="color:#2a9d8f; margin:0;">

<p style="margin:0 0 0 20px;">Syntax: <code>df['new_column'] = pd.Series(dictionary)</code></p>
<p style="margin:0 0 0 20px;">Example: <code>df['City'] = pd.Series({0:'New York',1:'Los Angeles',2:'Chicago'})</code></p>

<p style="color:#264653;"><b>The index in a DataFrame is crucial for data alignment and access. Here are some common operations for working with indices:</b></p>

1. <span style="color:#2a9d8f;">**Set Index (set column as Index)**</span>  
   `df.set_index('column_name', inplace=True)`  
   Replace 'column_name' with the name of the column to be set as the index.

2. <span style="color:#2a9d8f;">**Reindex (Reindex rows or columns)**</span>  
   `df.reindex(new_index)` for rows  
   `df.reindex(columns=new_columns)` for columns.  
   Provide `new_index` or `new_columns` with the new index or column labels.

3. <span style="color:#2a9d8f;">**Reset Index (Reset the index to default)**</span>  
   `df.reset_index(drop=False, inplace=True)`  
   Use `drop=True` to remove the index column from the DataFrame.
