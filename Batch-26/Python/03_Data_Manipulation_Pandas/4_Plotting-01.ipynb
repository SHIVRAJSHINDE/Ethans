# COUNT PLOT 

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(10,4))
ax = sns.countplot(x=df['Airline'])
ax.bar_label(ax.containers[0], fmt='%d', fontsize=8)
plt.xticks(rotation=90)
plt.show()
```

```python
df.head(2)
# Airline, Source, Destination, Route, Total_Stops
```

```python
# Count plot Destination
plt.figure(figsize=(8,4))
ax = sns.countplot(data=df, x='Destination')
ax.bar_label(ax.containers[0], fmt='%d', label_type='edge', fontsize=8)
plt.xticks(rotation=30)
plt.show()
```

```python
# Count plot Source

plt.figure(figsize=(10,4))
ax=sns.countplot(x=df['Source'],width=0.5)
ax.bar_label(ax.containers[0],fmt='%d',fontsize=8)
plt.xticks(rotation=90)
plt.show()
```

```python
# Create count plot
plt.figure(figsize=(10,5))
ax = sns.countplot(x=df['Total_Stops'], width=.5)
ax.bar_label(ax.containers[0], fmt='%d', fontsize=8)
plt.xticks(rotation=45)
plt.show()
```

# Line plot

```python
plt.figure(figsize=(10,3))
sns.lineplot(x=df['Airline'],y=df['Price'],marker='o')

plt.title("Ariline Vs Price")
plt.xticks(rotation=90)
plt.show()
```

```python
plt.figure(figsize=(12, 6))
sns.countplot(x=df['Airline'], order=df['Airline'].value_counts().index)
plt.title('Number of Flights by Airline', fontsize=16)
plt.xlabel('Airline')
plt.ylabel('Flight Count')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


```

```python
plt.figure(figsize=(12, 6))
sns.lineplot(x=df['Airline'], y=df['Price'], marker='o', errorbar=None)
plt.title('Price Trend Across Airlines', fontsize=16)
plt.xlabel('Airline')
plt.ylabel('Price')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(x=df['Price'], y=df['Proffit'],
                hue=df['Airline'], alpha=0.7, s=80)
plt.title('Relationship Between Price and Profit', fontsize=16)
plt.xlabel('Price')
plt.ylabel('Profit')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()


```

```python
plt.figure(figsize=(12, 6))
sns.boxplot(x=df['Airline'], y=df['Price'], hue=df['Airline'],
            legend=False)
plt.title('Price Distribution by Airline', fontsize=16)
plt.xlabel('Airline')
plt.ylabel('Price')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


```

```python
plt.figure(figsize=(12, 6))
sns.violinplot(x=df['Airline'], y=df['Price'], hue=df['Airline'],
               legend=False, inner='quartile')
plt.title('Price Distribution Density by Airline', fontsize=16)
plt.xlabel('Airline')
plt.ylabel('Price')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


```

```python
plt.figure(figsize=(12, 6))
sns.swarmplot(x=df['Airline'], y=df['Price'], size=4)
plt.title('Individual Flight Prices by Airline', fontsize=16)
plt.xlabel('Airline')
plt.ylabel('Price')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
