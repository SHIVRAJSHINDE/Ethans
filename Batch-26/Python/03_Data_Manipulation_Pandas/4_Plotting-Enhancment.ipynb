```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_excel("E:/1_Eathans/Ethans/Data/Airline/Swarn_Airline.xlsx")
df.head(2)
```

```python
order = df.groupby('Airline')['Price'].sum().sort_values().index
order
```

```python
df['Date_of_Journey'] = pd.to_datetime(df['Date_of_Journey'])
df['Date_of_Journey'].dt.month_name()
```

```python
sns.set_theme(style="dark",palette='pastel')
plt.figure(figsize=(10,5))

Airline_Sort = df.groupby('Airline')['Price'].sum().sort_values().index

ax = sns.barplot(x=df['Airline'],y=df['Price'],errorbar=None,estimator='sum',order=Airline_Sort,legend=False,alpha=.9,gap=.1)

ax.bar_label(ax.containers[0],rotation=90)

plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
sns.set_theme(style="whitegrid", palette='viridis')
plt.figure(figsize=(10,5))

AirLine_sort = df.groupby('Airline')['Price'].count().sort_values().index

ax=sns.countplot(x=df['Airline'],order=AirLine_sort, alpha=0.8, gap=0.3,legend=True)

ax.bar_label(ax.containers[0])
print(ax.containers[0])
plt.xticks(rotation=90)
plt.xlabel("Airline",fontsize=20)
plt.tight_layout()
plt.show()
```

```python
sns.set_theme(style="whitegrid", palette='Set1')
plt.figure(figsize=(10,5))

sns.lineplot(x=df['Airline'],y=df['Price'],errorbar=None,marker="s",estimator='mean',alpha=0.9,linestyle="-.")

plt.xticks(rotation=90)
plt.xlabel("Airline",fontsize=20)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,10))
sns.scatterplot(data=df,x='Price',y='Proffit',size='Airline',sizes=(40, 220),hue='Airline',palette="Set2",)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.boxplot(data=df,y='Price',gap=0.08,flierprops={"marker": "s", "markersize": 10, "alpha": 0.5})
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.violinplot(data=df,x='Price',inner="stick",bw_adjust=0.1)
```

```python
plt.figure(figsize=(15,6))
sns.swarmplot(data=df,hue='Source',y='Price',dodge=True,palette='Set2',size=4)
plt.show()
```
