# type casting in array or type coversion


```python
# astype
# if you want change the data type
import numpy as np

array_2d = np.array([[1, 2, 3,6], 
                     [4, 5, 6,8], 
                     [7, 8, 9,9]])
print(array_2d)
print(array_2d.dtype)
print("--------------")
d2Arr = array_2d.astype('float')
print(d2Arr)
print(d2Arr.dtype)
print("------------------")

d2Arr_2 = d2Arr.astype('int')
print(d2Arr_2)
print(d2Arr_2.dtype)
print("------------------")
```
