<h3 style="color: #2a9d8f; font-size: 18px; margin-top: 0; margin-bottom: 8px; font-weight: bold;">Syntax</h3>
<pre style="background-color: #ffffff; border: 1px solid #e1e8e7; padding: 10px; border-radius: 4px; font-family: monospace; font-size: 14px; margin-bottom: 20px; color: #264653;">pd.merge(df1, df2, left_on='name1', right_on='name2', how='outer')</pre>

<h3 style="color: #2a9d8f; font-size: 18px; margin-top: 0; margin-bottom: 10px; font-weight: bold;">Join Types</h3>
<ul style="font-size: 16px; line-height: 1.6; padding-left: 0; margin: 0; list-style-type: none; color: #264653;">
<li style="margin-bottom: 2px;"><strong style="color: #2a9d8f;"><code>how='outer'</code></strong> — Returns all rows from both DataFrames, filling missing values with NaN.</li>
<li style="margin-bottom: 2px;"><strong style="color: #2a9d8f;"><code>how='inner'</code></strong> — Returns only rows with matching keys in both DataFrames.</li>
<li style="margin-bottom: 2px;"><strong style="color: #2a9d8f;"><code>how='left'</code></strong> — Returns all rows from left DataFrame, with matching rows from right.</li>
<li style="margin-bottom: 0;"><strong style="color: #2a9d8f;"><code>how='right'</code></strong> — Returns all rows from right DataFrame, with matching rows from left.</li>
</ul>
</div>

```python
#  Create DataFrame
import pandas as pd
data1 = [("Alice", 1),("Charlie", 2),("Bob", 3)]
columns1 = ["name1", "value1"]

left_Df = pd.DataFrame(data1, columns=columns1)
left_Df

print()

data2 = [("Alice", 10), ("Bob", 20), ("David", 30)]
columns2 = ["name2", "value2"]
right_Df = pd.DataFrame(data2, columns=columns2)
right_Df

# pd.merge(leftDF,rightDf,left_on="ColunName",rightOn=ColumnName,how="left")

pd.merge(left_Df,right_Df,left_on="name1",right_on='name2',how="outer")

```

```python
# merge()
# Outer join
# in merge method common_key is considered

# pd.merge(df1, df2, left_on='NameOfCol', right_on='NameOfCol', how='outer')

pd.merge(df1,df2,left_on='name1',right_on='name2',how='outer')

```
