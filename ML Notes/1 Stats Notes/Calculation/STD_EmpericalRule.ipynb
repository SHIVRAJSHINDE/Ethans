```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
newDict ={'Values': [10,20,20,20,20,20,20,30,30,30,30,30,30,40,40,40,40,40,40,50,]}
df = pd.DataFrame(newDict)
# Mean and Standard Deviation
mean = df['Values'].mean()

mean= float(mean)
mean

std = df['Values'].std()
std = float(std)
std

print("Mean =", mean)
print("Standard Deviation =", std)
```

```python
# KDE Plot

plt.figure(figsize=(6,3))
sns.kdeplot(df['Values'], fill=True)
# Mean
plt.axvline(mean,linestyle='--')

plt.axvline(mean - std,linestyle=':',color='orange',label=f'{std:.2f}')
plt.axvline(mean + std,linestyle=':',color='orange',label=f'{std:.2f}')

plt.title("Empirical Distribution")
plt.legend()
plt.show()
```
